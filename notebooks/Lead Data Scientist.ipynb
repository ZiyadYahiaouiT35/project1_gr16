```python
############## Acquisition et d'Initialisation des données #################


import pandas as pd
import urllib.request
import os

# 1. Créer le dossier data s'il n'existe pas
if not os.path.exists('../data'):
    os.makedirs('../data')

# 2. L'adresse du fichier sur internet
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00519/heart_failure_clinical_records_dataset.csv"

# 3. Destination sur ton ordinateur
destination = "../data/heart.csv"

# 4. Téléchargement magique
try:
    urllib.request.urlretrieve(url, destination)
    print("Succès ! Le fichier est maintenant dans ton dossier /data")
    
    # 5. Vérification : On l'ouvre pour voir
    df = pd.read_csv(destination)
    display(df.head())
except Exception as e:
    print(f"Erreur : {e}")
```

```python
# Donne la moyenne, le min, le max pour toutes les colonnes d'un coup
df.describe()
```

```python
################## Diagnostic du déséquilsibre des classes ###################
import seaborn as sns
import matplotlib.pyplot as plt

# Création d'un graphique à barres pour compter les morts (1) et les survivants (0)
plt.figure(figsize=(8, 5))
sns.countplot(x='DEATH_EVENT', data=df, palette='viridis')

# Ajout de titres pour que ce soit professionnel
plt.title('Répartition des décès (0 = Survie, 1 = Décès)')
plt.xlabel('Événement de décès')
plt.ylabel('Nombre de patients')
plt.show()
```

```python
######Importation de l'algorithme d'échantillonnage synthétique (SMOTE)######
from imblearn.over_sampling import SMOTE
from collections import Counter

# 1. On sépare les caractéristiques (X) de la cible (y)
X = df.drop('DEATH_EVENT', axis=1)
y = df['DEATH_EVENT']

print(f"Avant SMOTE : {Counter(y)}")

# 2. On applique SMOTE
smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X, y)

print(f"Après SMOTE : {Counter(y_res)}")
```

```python
####### Vérification de la colinéarité et analyse de redondance ########
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
# On utilise df (le dataset original) pour voir les corrélations naturelles
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matrice de Corrélation (Analyse de Redondance)')
plt.show()
```

```python
############## Validation visuelle de l'équilibrage final ###############
import seaborn as sns
import matplotlib.pyplot as plt

# On transforme y_res en DataFrame pour l'affichage
df_equilibre = pd.DataFrame(y_res, columns=['DEATH_EVENT'])

plt.figure(figsize=(7, 5))

# On affiche la répartition finale
sns.countplot(x='DEATH_EVENT', data=df_equilibre, palette='Set2')

# Titre professionnel pour ton rapport
plt.title('Distribution des classes dans la nouvelle Dataset (Équilibrée)')
plt.xlabel('Issue Clinique (0 = Survie, 1 = Décès)')
plt.ylabel('Nombre total de patients')

# Ajout des étiquettes (203 sur chaque barre)
for i in range(2):
    count = len(df_equilibre[df_equilibre['DEATH_EVENT'] == i])
    plt.text(i, count + 2, str(count), ha='center', fontweight='bold')

plt.show()
```

```python
##################### Optimizing Memory #####################
# --- ÉTAPE 1 : ÉTAT INITIAL ---
print("📊 ANALYSE DE LA MÉMOIRE AVANT OPTIMISATION")
print("-" * 50)

# Affichage des types actuels
print("Types de données actuels :")
print(df.dtypes)

# Calcul de l'espace occupé en KB
mem_initiale = df.memory_usage(deep=True).sum() / 1024
print("-" * 50)
print(f"Espace mémoire occupé : {mem_initiale:.2f} KB")
```

```python
# --- ÉTAPE 2 : LA FONCTION D'OPTIMISATION ---
import pandas as pd

def optimize_memory(df):
    """
    Réduit la taille des données en ajustant les types numériques.
    """
    for col in df.columns:
        col_type = df[col].dtype
        
        # Optimisation pour les nombres entiers
        if str(col_type).startswith('int'):
            df[col] = pd.to_numeric(df[col], downcast='integer')
        
        # Optimisation pour les nombres décimaux
        elif str(col_type).startswith('float'):
            df[col] = pd.to_numeric(df[col], downcast='float')
            
    return df
```

```python
# --- ÉTAPE 3 : RÉSULTAT FINAL ---
# Application de la fonction
df_optimise = optimize_memory(df)

print("🚀 RÉSULTAT APRÈS OPTIMISATION")
print("-" * 50)

# Vérification des nouveaux types (int8, int16, float32, etc.)
print("Nouveaux types de données :")
print(df_optimise.dtypes)

# Calcul du gain
mem_finale = df_optimise.memory_usage(deep=True).sum() / 1024
gain = ((mem_initiale - mem_finale) / mem_initiale) * 100

print("-" * 50)
print(f"Nouvel espace mémoire : {mem_finale:.2f} KB")
print(f"Réduction totale : {gain:.1f} %")
```
